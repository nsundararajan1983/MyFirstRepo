## 1. Getting and Knowing your data

```python
import pandas as pd
#pd.read_csv(filepath_or_buffer="school_students_test_data_100.csv")
student_marks = pd.read_table(filepath_or_buffer="school_students_test_data_100.csv",sep=",")
student_marks["Cut_off"] = (student_marks["Maths"]+student_marks["Science"])/2
#student_marks[["Student_Name","Cut_off","Average"]].sort_values(by= "Cut_off", ascending = False).filter_values(["Cut_off"]>89)
student_marks.loc[(student_marks["Cut_off"]> 85) & (student_marks["Average"]>70) , ["Student_ID","Student_Name", "Cut_off","Average"]].sort_values(by="Cut_off", ascending=False).head(10)
```
