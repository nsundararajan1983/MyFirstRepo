## 1.4 Data visualization

### 1.4.1 Singlevariant Analysis

```python
from matplotlib import pyplot as plt
plt.hist(car_data["mpg"])
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.title("MPG - Millege Per Gallon")
plt.show()
```

#### Visualization insights
More sales occured on cars with millege 20-25 - this is place where company can focus to increase production.
Sales of cars with Millege less than 6is very low. this can be deprioritized, or marketting stratergies can be changed
